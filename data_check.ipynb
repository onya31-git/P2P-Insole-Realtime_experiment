```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# ---------------------------------------------------------
# 設定：調査対象のファイルパス（train.pyで問題が疑われるデータ）
# ---------------------------------------------------------
target_pairs = [
    (
        # Case 1: T004S003
        './data/training_data/Skeleton/T004S003_skeleton.csv',
        './data/20250518test4/InsoleSensor/3_final/T004S003_Insole_l.csv',
        './data/20250518test4/InsoleSensor/3_final/T004S003_Insole_r.csv',
    ),
    (
        # Case 2: T004S004
        './data/training_data/Skeleton/T004S004_skeleton.csv',
        './data/20250518test4/InsoleSensor/3_final/T004S004_Insole_l.csv',
        './data/20250518test4/InsoleSensor/3_final/T004S004_Insole_r.csv'
    )
]

# ---------------------------------------------------------
# ユーティリティ関数
# ---------------------------------------------------------
def read_pressure_csv(path: str) -> pd.DataFrame:
    """train.pyと同様の読み込みロジック"""
    df = pd.read_csv(path, skiprows=1, low_memory=False)
    if "Timestamp" in df.columns:
        df = df.drop(columns=["Timestamp"])
    # 数値変換（エラー回避のためcoerce）
    df = df.apply(pd.to_numeric, errors='coerce')
    return df

def analyze_dataset(skeleton_path, left_path, right_path):
    print(f"\n{'='*60}")
    print(f"Analyzing Dataset: {Path(skeleton_path).stem}")
    print(f"{'='*60}")
    
    # 1. データ読み込み
    try:
        skel = pd.read_csv(skeleton_path)
        left = read_pressure_csv(left_path)
        right = read_pressure_csv(right_path)
    except FileNotFoundError as e:
        print(f"Error: File not found - {e}")
        return

    # データ長チェック
    min_len = min(len(skel), len(left), len(right))
    print(f"Data Lengths -> Skeleton: {len(skel)}, Left: {len(left)}, Right: {len(right)}")
    print(f"Effective Training Length (min): {min_len}")

    # 2. SkeletonデータのNaN詳細分析
    print(f"\n--- Skeleton Data Analysis ---")
    
    # 全体のNaN数
    total_nans = skel.isna().sum().sum()
    print(f"Total NaNs in Skeleton: {total_nans}")

    # 「全ての行がNaN」である列（マーカー）を探す
    # これが存在すると、ffill/bfillでも埋まらず、学習がクラッシュする原因になります
    all_nan_cols = skel.columns[skel.isna().all()].tolist()
    
    if all_nan_cols:
        print(f"\n[CRITICAL ERROR] Found columns that are ALL NaN (cannot be interpolated):")
        print(f"Count: {len(all_nan_cols)}")
        print(f"Columns: {all_nan_cols}")
    else:
        print("\n[OK] No columns are completely NaN.")

    # train.pyの補間ロジックをシミュレーション
    skel_filled = skel.fillna(method='bfill').fillna(method='ffill')
    remaining_nans = skel_filled.isna().sum().sum()
    
    if remaining_nans > 0:
        print(f"\n[WARNING] After ffill/bfill, {remaining_nans} NaNs still remain!")
        print("This confirms that simple interpolation is insufficient for this data.")
    else:
        print("\n[OK] ffill/bfill successfully removed all NaNs.")

    # 3. インソールデータのチェック
    print(f"\n--- Insole Data Analysis ---")
    print(f"Left Insole NaNs: {left.isna().sum().sum()}, Infs: {np.isinf(left).sum().sum()}")
    print(f"Right Insole NaNs: {right.isna().sum().sum()}, Infs: {np.isinf(right).sum().sum()}")

    # 4. 可視化 (NaNの位置をヒートマップで表示)
    if total_nans > 0:
        plt.figure(figsize=(12, 6))
        sns.heatmap(skel.isna(), cbar=False, cmap='viridis', yticklabels=False)
        plt.title(f"Missing Values Map (Yellow=NaN): {Path(skeleton_path).name}")
        plt.xlabel("Markers (Columns)")
        plt.ylabel("Time (Frames)")
        plt.show()

# ---------------------------------------------------------
# 実行
# ---------------------------------------------------------
for s_path, l_path, r_path in target_pairs:
    analyze_dataset(s_path, l_path, r_path)
```
